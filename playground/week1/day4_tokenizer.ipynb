# Tokenizing with Code

`tiktoken` is a fast Byte Pair Encoding (BPE) tokenizer developed by OpenAI specifically for use with their large language models. It allows you to programmatically count tokens, encode text into token IDs, and decode those IDs back into text.

Because OpenAI bills API requests based on token usage and enforces maximum context windows, tiktoken is a crucial tool for **estimating API costs** and **managing your context window limits** before hitting the server

https://platform.openai.com/tokenizer

```python
import tiktoken
```

```python
gpt_tokenizer = tiktoken.encoding_for_model('gpt-5-turbo')
tokens = gpt_tokenizer.encode("Hi my name is Ed and I like banoffee pie")
tokens
```

```python
for token in tokens:
    token_text = gpt_tokenizer.decode([token])
    print(f"{token} = '{token_text}'")
```
